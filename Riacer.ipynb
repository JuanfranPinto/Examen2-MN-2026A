```python
import numpy as np
from scipy import optimize

# polonomio
def f(x):
    return x**5 + 5.25*x**4 + 4.125*x**3 - 9.125*x**2 - 14.625*x - 5.625

## el metodo de la biseccion 
def encontrar_raices():
    raices = []
    
    puntos = np.linspace(-10, 10, 400)

    for i in range(len(puntos) - 1):
        a = puntos[i]
        b = puntos[i + 1]

        if f(a) * f(b) < 0:
            try:
                raiz = optimize.bisect(
                    f,
                    a,
                    b,      
                    xtol=1e-10,
                    maxiter=100
                )

                raiz = float(raiz)

                # Evitar raíces repetidas
                if all(abs(raiz - r) > 1e-5 for r in raices):
                    raices.append(raiz)

            except:
                pass
                
        elif f(a) == 0:
            if all(abs(a - r) > 1e-5 for r in raices):
                raices.append(float(a))

    raices.sort()

    while len(raices) < 6:
        raices.append(np.nan)

    return raices[:6]

resultado = encontrar_raices()

print("raices:\n")

for i, r in enumerate(resultado, start=1):
    if np.isnan(r):
        print(f"x{i} = NaN")
    else:
        print(f"x{i} = {r:.10f}")
```
